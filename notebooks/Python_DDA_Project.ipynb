## Outlier Detection and Visualization

```python
# Detecting outliers using IQR method

def outlier_limits(data, column):
    Q1 = data[column].quantile(0.25)
    Q3 = data[column].quantile(0.75)
    #calculate middle 50% 
    IQR = Q3 - Q1
    #calculate lowwer and upper limit for detectiong outliners
    lower_limit = Q1 - 1.5 * IQR
    upper_limit = Q3 + 1.5 * IQR

    return lower_limit, upper_limit


# Check outliers in Quantity, Price and TotalAmount
for column in ["Quantity", "Price", "TotalAmount"]:
    lower1, upper1 = outlier_limits(df1_sales, column)
    lower2, upper2 = outlier_limits(df2_sales, column)

    print(f"\n{column} - 2009-2010")
    print("Lower limit:", round(lower1, 2))
    print("Upper limit:", round(upper1, 2))

    print(f"{column} - 2010-2011")
    print("Lower limit:", round(lower2, 2))
    print("Upper limit:", round(upper2, 2))
```

```python
# Count the number of outliers

for column in ["Quantity", "Price", "TotalAmount"]:
    lower1, upper1 = outlier_limits(df1_sales, column)
    lower2, upper2 = outlier_limits(df2_sales, column)

    # Count values outside the limits
    outliers1 = ((df1_sales[column] < lower1) | 
                 (df1_sales[column] > upper1)).sum()
    
    outliers2 = ((df2_sales[column] < lower2) | 
                 (df2_sales[column] > upper2)).sum()

    print(f"\n{column}")
    print("Outliers in 2009-2010:", outliers1)
    print("Outliers in 2010-2011:", outliers2)
```

### Visualize outliers using boxplots

```python
# Visualize outliers using boxplots

for column in ["Quantity", "Price", "TotalAmount"]:
    
    plt.figure(figsize=(7, 5))

    plt.boxplot(
        [df1_sales[column], df2_sales[column]],
        tick_labels=["2009-2010", "2010-2011"]
    )

    plt.title(f"{column} - Outlier Detection")
    plt.xlabel("Year")
    plt.ylabel(column)

    plt.grid(axis="y", alpha=0.3)

    plt.tight_layout()
    plt.show()
```
